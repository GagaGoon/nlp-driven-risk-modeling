```python
import numpy as np
import pandas as pd
import yfinance as yf
```

**Daily return**
$$r_t = \frac{P_t}{P_{t-1}} - 1$$

```python
def asset_return(close, period):
    return close.pct_change(fill_method=None, periods=period).iloc[period:]
```

```python
tickers = ['AAPL', 'MSFT', 'AMZN', 'JPM', 'XOM', 'JNJ', 'WMT', 'CAT', 'NEE', 'KO',
           'UPS', 'NVDA']
start_date = "1999-11-10"
end_date = "2026-01-01"

data = yf.download(
    tickers=tickers,
    start=start_date,
    end=end_date,
    interval="1d",
    auto_adjust=False,
    group_by="column",
    progress=False,
)

if data is None:
    raise TypeError('OHLCV not downloaded')

data = data.drop(columns='Close')
data.columns
# if data is None or data.empty:
#         raise ValueError("Не удалось получить цены.")

# close = data["Adj Close"].reindex(columns=tickers) # type: ignore
# close.index.name = "Date"
# close.columns.name = None

# close.head()
```

```python
returns = asset_return(data['Adj Close'], 5)

returns.head()
```

```python
data = pd.concat(
    [data, pd.concat({"Returns": returns}, axis=1)],
    axis=1
)
data = data[5:]
data
```
