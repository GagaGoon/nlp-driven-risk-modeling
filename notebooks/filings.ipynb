# Парсинг risk factors из годового отчета Apple

```python
from edgar import Company

company = Company("AAPL")
filings = company.get_filings(form="10-K")
filing = filings.latest()

print(filing.company)
print(filing.form)
print(filing.filing_date)
```

```python
filing.view()
```

```python
tenk = filing.obj()

risk_factors = tenk.risk_factors

print(risk_factors)
```

# Парсинг Risk Factors из 12 компаний, за 10-K, 10-Q за последний год

```python
companies = ['AAPL', 'MSFT', 'AMZN', 'JPM', 'XOM', 'JNJ', 'WMT', 'CAT', 'NEE', 'KO',
             'UPS', 'NVDA']


for company in companies:
    filings = company.get_filings(form="10-K")
    filing = filings.latest()
    tenk = filing.obj()
    risk_factors = tenk.risk_factors

    filename = company
    with open("text.txt", "w", encoding="utf-8") as file:
        file.write(risk_factors)
```
