```python
from pathlib import Path

import pandas as pd
```

Finding if there was extraction fails in filings since start of 2006

# Only for AAPL

```python
project_dir = Path.cwd().parent.parent
indexes_path = project_dir / 'data' / 'indexes' / 'extraction'
aapl_index = indexes_path / 'AAPL.csv'
extractions = pd.read_csv(aapl_index, dtype=str, keep_default_na=False)
extractions
```

```python
extractions_since_2006 = extractions[extractions['period_ended'] > '2005-12-31']
extractions_since_2006
```

```python
failed_extractions = extractions_since_2006[extractions_since_2006['extraction_status']
                                            != 'extracted']
failed_extractions
```

```python
count = len(failed_extractions)
count
```

# For all compaines

```python
for index in sorted(indexes_path.glob("*.csv")):
    ticker = index.stem

    extractions = pd.read_csv(index, dtype=str, keep_default_na=False)
    extractions_since_2006 = extractions[extractions['period_ended'] > '2005-12-31']
    failed_extractions = extractions_since_2006[
        extractions_since_2006['extraction_status'] != 'extracted'
        ]
    count = len(failed_extractions)

    print(ticker, count)
```
