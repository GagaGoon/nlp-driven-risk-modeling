## 2 problem

In the **table of contents** under the `Item 1A *` there are
`* Item omitted because no answer is called for or item is not applicable.`

```python
import re
```

```python
def explicitly_omitted_in_toc(filing):
    # Сохраняем переносы строк для определения начала основного текста.
    text = filing.text()

    toc_start = re.search(
        r"Table\s+of\s+Contents\s*"
        r"(?=Part\s+I\.?\s+Financial\s+Information)",
        text,
        re.IGNORECASE,
    )
    if not toc_start:
        return False

    # Два самостоятельных заголовка обозначают начало основного текста.
    body_pattern = re.compile(
        r"^\s*Part\s+I\.?\s+Financial\s+Information\s*\n"
        r"\s*Item\s+1\.?\s+Financial\s+Statements\s*$",
        re.IGNORECASE | re.MULTILINE,
    )
    body_start = body_pattern.search(text, toc_start.end())
    if not body_start:
        return False

    # Сноска должна находиться внутри оглавления.
    note_pattern = re.compile(
        r"\*\s*Item omitted because no answer is called for "
        r"or item is not applicable\.",
        re.IGNORECASE,
    )
    note = note_pattern.search(text, toc_start.end(), body_start.start())
    if not note:
        return False

    # Пункт со звёздочкой должен находиться перед сноской.
    item_pattern = re.compile(
        r"Item\s+1A\.?\s+Risk\s+Factors\s*\*",
        re.IGNORECASE,
    )
    return bool(item_pattern.search(text, toc_start.end(), note.start()))


accession_numbers = failed_extractions['accession_number']

for accession_number in accession_numbers:
    file_path = '../../data/raw/10-Q/CAT/' + accession_number + '.txt'
    filing = LocalFiling.from_sgml(file_path)
    status = (
        "explicitly_omitted"
        if explicitly_omitted_in_toc(filing)
        else "not_found"
    )
    failed_extractions.loc[
        failed_extractions['accession_number'] == accession_number,
        'extraction_status',
    ] = status
```
